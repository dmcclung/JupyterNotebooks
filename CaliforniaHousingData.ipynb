```python
from sklearn.model_selection import train_test_split
```

```python
X = housing.data  # Features
y = housing.target  # Target values

# Split the data (default is 75% train, 25% test)
X_train_raw, X_test_raw, y_train, y_test = train_test_split(X, y, random_state=42)
```

```python
X_train_mean = np.mean(X_train_raw, axis=0)
X_train_mean
```

```python
X_train_std = np.std(X_train_raw, axis=0)
X_train_std
```

```python
X_train = X_train_raw - X_train_mean / X_train_std
X_train
```

```python
X_test = X_test_raw - X_train_mean / X_train_std
X_test
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset
```

```python
X_train_tensor = torch.FloatTensor(X_train)
X_train_tensor
```

```python
print(X_train.dtype)  # NumPy array type
print(X_train_tensor.dtype)  # PyTorch tensor type
```

```python
X_test_tensor = torch.FloatTensor(X_test)
X_test_tensor
```

```python
y_train_tensor = torch.FloatTensor(y_train.reshape(-1, 1))
y_train_tensor
```

```python
y_train
```

```python
y_test_tensor = torch.FloatTensor(y_test.reshape(-1, 1))
y_test_tensor
```

```python
y_test
```

```python
train_dataset = TensorDataset(X_train_tensor, y_train_tensor)
train_dataset
```

```python
train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
train_loader
```

```python
 model = nn.Sequential(
     nn.Linear(8, 64),
     nn.ReLU(),
     nn.Linear(64, 32),
     nn.ReLU(),
     nn.Linear(32, 16),
     nn.ReLU(),
     nn.Linear(16,1),
 )
```

```python
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

```python
num_epochs = 200
```

```python
for epoch in range(num_epochs):
    model.train()
    running_loss = 0.0
    
    for inputs, targets in train_loader:
        # Zero the parameter gradients
        optimizer.zero_grad()
        
        # Forward pass
        outputs = model(inputs)
        loss = criterion(outputs, targets)
        
        # Backward pass and optimize
        loss.backward()
        optimizer.step()
        
        running_loss += loss.item()
    
    # Print statistics
    if (epoch+1) % 10 == 0:  # Print every 10 epochs
        print(f'Epoch {epoch+1}/{num_epochs}, Loss: {running_loss/len(train_loader):.4f}')

# Evaluate the model
model.eval()
with torch.no_grad():
    test_predictions = model(X_test_tensor)
    test_loss = criterion(test_predictions, y_test_tensor)
    print(f'Test MSE: {test_loss:.4f}')
    
    # Calculate R^2 score
    y_mean = torch.mean(y_test_tensor)
    ss_tot = torch.sum((y_test_tensor - y_mean) ** 2)
    ss_res = torch.sum((y_test_tensor - test_predictions) ** 2)
    r2 = 1 - (ss_res / ss_tot)
    print(f'R^2 Score: {r2:.4f}')
```
